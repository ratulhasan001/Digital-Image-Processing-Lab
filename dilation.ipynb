```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
image = cv2.imread('Downloads/img.jpg', cv2.IMREAD_GRAYSCALE)
x, y = map(int, input().split())
kernel = np.array([input().strip().split() for _ in range(x)], int)
_, binary_image = cv2.threshold(image, 0, 255, cv2.THRESH_BINARY + cv2.THRESH_OTSU)

def erosion(image, structuring_element):
    h = x // 2
    w = y // 2
    padded_image = np.pad(image, pad_width=2, mode='constant', constant_values=0)

    output_image = np.zeros_like(image)

    for i in range(h, padded_image.shape[0] - h):  
        for j in range(w, padded_image.shape[1] - w):  

            region = padded_image[i - h : i + h + 1, j - w : j + w + 1]

            erosion_val = np.max(region * structuring_element)
            output_image[i - h, j - w] = erosion_val

    return output_image


# kernel = cv2.getStructuringElement(cv2.MORPH_RECT,(5,5))  # auto
print("Kernel for Morphological Transformations")
print(kernel)
print('shape of binary image=',binary_image.shape)
eroded_image=erosion(binary_image,kernel)


img_erosion = cv2.erode(binary_image, kernel, iterations=1)
print('Eroded image by build in function to compare the result:')
plt.imshow(img_erosion,'gray')
plt.show()
print('Mannualy performed result:')
plt.imshow(eroded_image,'gray')
plt.show()

```
