```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
image = cv2.imread('Downloads/clock.jpg', cv2.IMREAD_GRAYSCALE)

kernel_horizontal = np.array([[-1, -1, -1],
                            [2, 2, 2],
                           [-1, -1, -1]])
kernel_vertical = np.array([[-1, 2, -1],
                           [-1, 2, -1],
                           [-1, 2, -1]])
line_horizontal = cv2.filter2D(image,-1,kernel_horizontal)
line_vertical = cv2.filter2D(image,cv2.CV_64F,kernel_vertical)
plt.subplot(1,2,1)
plt.imshow(line_horizontal,cmap='gray')
plt.title("Horizontal Lines")
plt.axis("off")
plt.show()
plt.subplot(1,2,2)
plt.imshow(line_vertical,cmap='gray')
plt.title("Vertical Lines")
plt.axis("off")
plt.show()
```

```python
kernel_roberts_x = np.array([
    [1, 0],
    [0, -1]
])
kernel_roberts_y = np.array([
    [0, 1],
    [-1, 0]
])
edge_roberts_x = cv2.filter2D(image,cv2.CV_64F,kernel_roberts_x)
edge_roberts_y = cv2.filter2D(image,cv2.CV_64F,kernel_roberts_y)
edge_roberts = np.sqrt(edge_roberts_x**2 + edge_roberts_y**2)

plt.imshow(edge_roberts, cmap='gray')
plt.title("Robert Edge Detection")
plt.axis("off")
plt.show()
```

```python
Prewitt_x = np.array([
    [-1,-1,-1],
    [0,0,0],
    [1,1,1],
])
Prewitt_y = np.array([
    [-1,0,1],
    [-1,0,1],
    [-1,0,1]
])
Prewitt_edge_x = cv2.filter2D(image,cv2.CV_64F,Prewitt_x)
Prewitt_edge_y = cv2.filter2D(image,cv2.CV_64F,Prewitt_y)
Prewitt_edge = np.sqrt(Prewitt_edge_x**2 + Prewitt_edge_y**2)

plt.imshow(Prewitt_edge, cmap='gray')
plt.title("Prewitt Edge Detection")
plt.axis("off")
plt.show()
```

```python
sobel_x = np.array([[-1, -2, -1],
                       [0, 0, 0],
                       [1, 1, 1]])

sobel_y = np.array([[-1, 0, 1],
                       [-2, 0, 2],
                       [-1, 0, 1]])

sobel_edge_x = cv2.filter2D(image,cv2.CV_64F,sobel_x)
sobel_edge_y = cv2.filter2D(image,cv2.CV_64F,sobel_y)
sobel_edge = np.sqrt(sobel_edge_x**2 + sobel_edge_y**2)

plt.imshow(sobel_edge, cmap='gray')
plt.title("Sobel Edge Detection")
plt.axis("off")
plt.show()
```

```python
lap = np.array([[0, 1, 0],
                [1, -4, 1],
                [0, 1, 0]])


lap_edge = cv2.filter2D(image,cv2.CV_64F,lap)

plt.imshow(lap_edge, cmap='gray')
plt.title("Laplacian Edge Detection")
plt.axis("off")
plt.show()
```
