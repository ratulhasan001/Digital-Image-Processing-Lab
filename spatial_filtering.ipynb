```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
image = cv2.imread('Downloads/img.jpg', cv2.IMREAD_GRAYSCALE)
x, y = map(int, input().split())
kernel = np.array([input().strip().split() for _ in range(x)], int)

def average_filtering(image):
    output_image=np.zeros_like(image)
    sz_height = x // 2
    sz_width = y // 2
    for i in range(sz_height,image.shape[0] - sz_height):
        for j in range(sz_width,image.shape[1] - sz_width):
            region = image[i - sz_height:i + sz_height + 1, j - sz_width:j + sz_width + 1]
            av_fil_value=np.mean(region * kernel) # min, median, max
            output_image[i,j]=av_fil_value
    return output_image
smoothed_image=average_filtering(image)
plt.imshow(image,'gray')
plt.axis('off')
plt.title("Original Image")
plt.show()
plt.imshow(smoothed_image,'gray')
plt.axis('off')
plt.title("Smoothed Image")
plt.show()
```
